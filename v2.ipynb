## 1. Question Proposal and Hypothesis Definition

### Framing the Problem
**Context and Significance of the Research**  
Corruption remains a critical challenge across Europe, impacting governance, social equity, and public trust in institutions. While some countries experience high levels of visible corruption, others may have issues related to institutions or cultural norms. Understanding the relationship between perceived and actual corruption is important for implementing anti-corruption strategies and rebuilding public trust. 

**Initial Exploration of Hypotheses**  
The analysis focuses on the following key questions:
1. How well can the level of corruption of a country in Europe be quantified?
2. What differences are there in actual corruption and perceived corruption?
3. Are there different forms of corruption prevalent in different countries in Europe?
4. What characteristics of a country predict the level of corruption?
5. What characteristics of a country predict an increase or decrease in
the level of corruption?
---

### Defining Hypotheses
1. **Hypothesis 1: Misalignment Between Perceived and Actual Corruption**  
   There is a measurable gap between perceived corruption (e.g., QB5: Perceived Widespread Corruption) and actual corruption (e.g., Tolerance Index to Corruption, CPI_avg). This gap varies significantly by country and sector.

4. **Hypothesis 2: Socio-Economic Indicators as Predictors**  
   Socio-economic factors such as `GDP_Per_Capita`, `life expectancy`, and `Unemployment_Rate` are key predictors of both perceived and actual corruption levels.

#### CPI Preprocessing

```python
def read_excel_cpi(filename: str, sheet_name: Union[str, int]):
    path = os.path.join("..", "input", "cpi-2019-2023", filename)
    return pd.read_excel(path, sheet_name=sheet_name)

def clean_rows(data: pd.DataFrame, from_row: int = 2):
    # CPI data seems to always have three useless rows at the top of the excel sheet.
    data = data.iloc[from_row:]
    # Make the first row the column names
    data.columns = data.iloc[0]
    # Drop the first row
    data = data.iloc[1:]
    # Reset the index
    data.reset_index(drop=True, inplace=True)
    return data

def enforce_type(data: pd.DataFrame):
    # Enforce all columns except the first 3 to be numeric
    for col in data.columns[3:]:
        data[col] = pd.to_numeric(data[col], errors="coerce")
    return data


def load_cpi(filename: str, sheetname: str, from_row: int = 1):
    return enforce_type(clean_rows(read_excel_cpi(filename, sheetname), from_row))


def load_all_cpi():
    return {
        2019: load_cpi("CPI2019-1.xlsx", "CPI2019"),
        2020: load_cpi("CPI2020_GlobalTablesTS_210125.xlsx", "CPI2020"),
        2021: load_cpi("CPI2021_GlobalResults_Trends.xlsx", "CPI 2021"),
        2022: load_cpi("CPI2022_GlobalResultsTrends.xlsx", "CPI 2022 (final)"),
        2023: load_cpi("CPI2023_Global_Results__Trends.xlsx", "CPI 2023", 2),
    }

def concat_cpis(dfs: dict):
    for _, df in dfs.items():
        for col in df.columns:
            if col.startswith("CPI score"):
                df.rename(columns={col: "CPI score"}, inplace=True)
            elif col.startswith("Country /"):
                df.rename(columns={col: "Country"}, inplace=True)
            elif col.startswith("standard error"):
                df.rename(columns={col: "Standard error"}, inplace=True)
            elif col.startswith("Global Insight"):
                df.rename(
                    columns={col: "Global Insights Country Risk Ratings"}, inplace=True
                )
            elif col.startswith("Freedom House Nations in Transit Ratings"):
                df.rename(
                    columns={col: "Freedom House Nations in Transit"}, inplace=True
                )

    # Add a new column for the year and place it between "Region" and "CPI Score"
    for year, df in dfs.items():
        df.insert(3, "Year", year)

    return pd.concat(dfs.values(), ignore_index=True)
```

#### Corruption Risk preprocessing

```python
def read_excel_corruption_risk(filename: str, sheet_name: Union[str, int]):
    path = os.path.join("..", "input", "corruptionrisk", filename)
    return pd.read_excel(path, sheet_name=sheet_name)

def renaming(df: pd.DataFrame):
    for colname in df.columns:
        col = colname.lower().replace(" ", "")
        if col.startswith("countryname"):
            df.rename(columns={colname: "Country"}, inplace=True)
        elif col.startswith("countrycode"):
            df.rename(columns={colname: "ISO3"}, inplace=True)
        elif col.startswith("ipi"):
            df.rename(columns={colname: "Public Integrity"}, inplace=True)
        elif col.startswith("region"):
            df.rename(columns={colname: "Region"}, inplace=True)


def cleaning_transparency(df: pd.DataFrame):
    # Drop the third column
    df.drop(columns=["Countrycode 2"], inplace=True)
    renaming(df)
    return df

def cleaning_public_integrity(df: pd.DataFrame, year: int):
    # Drop the first 9 rows
    df = df.iloc[9:]
    # Make the first row the header
    df.columns = df.iloc[0]
    # Drop the first row
    df = df.iloc[1:]
    # Add the year column
    df["Year"] = year
    renaming(df)

    if year == 2023:
        # From column 5 onwards, replace all "," by "."
        for col in df.columns[5:]:
            df[col] = df[col].apply(lambda x: str(x).replace(",", "."))

        # Divide the population column by 1e6
        df["population"] = df["population"].apply(lambda x: float(x) / 1e6)

        # Format urban population column:
        # drop last % sign and make it a float
        df["urban population"] = df["urban population"].apply(
            lambda x: float(str(x).replace("%", ""))
        )

    df = enforce_types(df)

    return df

def enforce_types(df: pd.DataFrame):
    # From column 5 onwards, enforce numeric type
    for col in df.columns[5:]:
        df[col] = pd.to_numeric(df[col], errors="coerce")
    return df

def missing_values_integrity(df: pd.DataFrame):
    # NOTE: Starting from the 2021 edition, administrative burden and trade
    # openness have been replaced by administrative transparency and online
    # services, due to unavailable alternative data in the original components
    # (based on the World Bank Doing Business project, which closed).
    # Source: https://corruptionrisk.org/ipi-methodology/

    # This means that missing values in these columns cannot be made up. Bad luck I guess

    # Mean years of schooling has missing data for 2023
    # Impute by extrapolation
    df.set_index(["ISO3", "Year"], inplace=True)

    # Extrapolate data using 2019 and 2022
    for iso3 in df.index.levels[0]:
        try:
            if pd.isna(df.loc[(iso3, 2023), "mean years of schooling"]):
                value_2019 = df.loc[(iso3, 2019), "mean years of schooling"]
                value_2022 = df.loc[(iso3, 2021), "mean years of schooling"]
                extrapolated_value = value_2022 + (value_2022 - value_2019) / 2
                df.loc[(iso3, 2023), "mean years of schooling"] = extrapolated_value
        except KeyError:
            # Not enough data to interpolate, these missing values will remain
            pass

    # Venezuela has data for 2019 regarding income class, but not for 2021 and 2023.
    # Simply copy the 2019 data to 2021 and 2023
    for year in [2021, 2023]:
        if pd.isna(df.loc[("VEN", year), "income_class "]):
            df.loc[("VEN", year), "income_class "] = df.loc[
                ("VEN", 2019), "income_class "
            ]
        if pd.isna(df.loc[("VEN", year), "income_class_shortcode"]):
            df.loc[("VEN", year), "income_class_shortcode"] = df.loc[
                ("VEN", 2019), "income_class_shortcode"
            ]

    # North Macedonia and DR Congo have missing values for urban population in
    # 2021 and 2023 respectively
    # Impute by linear interpolation/extrapolation
    df.loc[("MKD", 2021), "urban population"] = (
        df.loc[("MKD", 2019), "urban population"]
        + df.loc[("MKD", 2023), "urban population"]
    ) / 2

    df.loc[("COD", 2023), "urban population"] = (
        df.loc[("COD", 2021), "urban population"]
        - df.loc[("COD", 2019), "urban population"]
    ) / (2021 - 2019) * (2023 - 2019) + df.loc[("COD", 2019), "urban population"]

    # Despite the efforts, the data for these two countries (and maybe for more)
    # is botched. I can already they are outliers (at least in the population column)
    
    df.reset_index(inplace=True)
    return df


def concat_public_integrity(df: list):
    # Concatenate the public integrity index
    return pd.concat(df)
```

```python
QUESTION_DICT = {
    "QB1": "Visited Public Healthcare Practitioner (Last 12 Months)",
    "QB2": "Extra Payment or Gift to Public Healthcare Practitioner",
    "QB3": "Description of Extra Payment Incident",
    "QB4.1": "Acceptability of Giving Money (Public Services)",
    "QB4.2": "Acceptability of Giving a Gift (Public Services)",
    "QB4.3": "Acceptability of Doing a Favor (Public Services)",
    "QB4T": "Tolerance Index to Corruption",
    "QB5": "Perceived Widespread Corruption",
    "QB6": "Corruption Trend in the Last 3 Years",
    "QB7": "Corruption Widespread in Sectors",
    "QB8": "Knows Someone Who Takes Bribes",
    "QB10": "Knows Where to Report Corruption",
    "QB12": "Experienced or Witnessed Corruption",
    "QB_avg": "Average of All QB Questions",
    "QB6_QB13": "Correlation Between Reporting and Corruption Trend",
    "QB15.6_QB15.7": "Attitudes Towards Bribes (Money and Gifts)",
    "QB15.1": "Corruption Exists in Local/Regional Institutions",
    "QB15.2": "Corruption Exists in National Institutions",
    "QB15.3": "Corruption is Part of Business Culture",
    "QB15.4": "Personally Affected by Corruption in Daily Life",
    "QB15.5": "Sufficient Judicial Prosecution to Deter Corruption",
    "QB15.6": "High-Level Corruption Not Prosecuted Enough",
    "QB15.7": "Government Efforts Against Corruption are Effective",
    "QB15.8": "Close Ties Between Business and Politics Lead to Corruption",
    "QB15.9": "Corruption or Connections Ease Access to Public Services",
    "QB15.10": "Transparency and Oversight in Political Party Funding",
    "QB15.11": "Success in Business Requires Political Connections",
    "QB15.12": "Favoritism and Corruption Hinder Business Competition",
    "QB15.13": "Anti-Corruption Measures are Impartially Applied",
    "CPI_avg": "Average Corruption Perception Index (2020-2023)",
    "CPI_percent_change": "CPI Percent Change (2020-2023)",
    "Latitude": "Latitude (Country)",
    "Longitude": "Longitude (Country)"
}

def handle_missing_values(df, strategy='mean'):
    for col in df.columns:
        if df[col].dtype in ['float64', 'int64']:
            if strategy == 'mean':
                df[col] = df[col].fillna(df[col].mean())
            elif strategy == 'median':
                df[col] = df[col].fillna(df[col].median())
        else:
            df[col] = df[col].replace('- ', np.nan)
            df[col] = df[col].astype(str).fillna(df[col].mode()[0])
    return df

def encode_categorical(df, columns):
    label_encoders = {}
    for col in columns:
        df[col] = df[col].astype(str).str.strip().replace('- ', np.nan)
        df[col] = df[col].fillna('Unknown')
        le = LabelEncoder()
        encoded_col_name = f"{col}_ID"
        df[encoded_col_name] = le.fit_transform(df[col])
        label_encoders[col] = le
    return df, label_encoders

def normalize_features(df, columns):
    scaler = StandardScaler()
    df[columns] = scaler.fit_transform(df[columns])
    return df, scaler

def detect_outliers(df, columns):
    for col in columns:
        if df[col].dtype in ['float64', 'int64']:
            Q1 = df[col].quantile(0.25)
            Q3 = df[col].quantile(0.75)
            IQR = Q3 - Q1
            lower_bound = Q1 - 1.5 * IQR
            upper_bound = Q3 + 1.5 * IQR
            df[col] = np.where(df[col] < lower_bound, lower_bound, df[col])
            df[col] = np.where(df[col] > upper_bound, upper_bound, df[col])
    return df

def create_composite_indices(df):
    df['QB4T'] = df[['QB4.1', 'QB4.2', 'QB4.3']].mean(axis=1)
    return df

def add_coordinates(df):
    coordinates = {
        'BE': (50.8503, 4.3517), 'BG': (42.6977, 23.3219), 'CZ': (50.0755, 14.4378),
        'DK': (55.6761, 12.5683), 'DE': (51.1657, 10.4515), 'EE': (59.4370, 24.7536),
        'IE': (53.3498, -6.2603), 'EL': (37.9838, 23.7275), 'ES': (40.4168, -3.7038),
        'FR': (48.8566, 2.3522), 'HR': (45.8150, 15.9819), 'IT': (41.9028, 12.4964),
        'CY': (35.1264, 33.4299), 'LV': (56.9496, 24.1052), 'LT': (54.6872, 25.2797),
        'LU': (49.8153, 6.1296), 'HU': (47.4979, 19.0402), 'MT': (35.8997, 14.5146),
        'NL': (52.3676, 4.9041), 'AT': (48.2082, 16.3738), 'PL': (52.2297, 21.0122),
        'PT': (38.7169, -9.1390), 'RO': (44.4268, 26.1025), 'SI': (46.0569, 14.5058),
        'SK': (48.1486, 17.1077), 'FI': (60.1695, 24.9354), 'SE': (59.3293, 18.0686),
        'UK': (51.5074, -0.1278)
    }
    df['Latitude'] = df['Country'].map(lambda x: coordinates.get(x, (None, None))[0])
    df['Longitude'] = df['Country'].map(lambda x: coordinates.get(x, (None, None))[1])
    return df

def calculate_additional_features(df):
    qb_columns = [col for col in df.columns if col.startswith('QB')]
    if qb_columns:
        df['QB_avg'] = df[qb_columns].mean(axis=1)

    if 'QB6' in df.columns and 'QB13' in df.columns:
        df['QB6_QB13'] = df['QB6'] * df['QB13']

    if 'QB15.6' in df.columns and 'QB15.7' in df.columns:
        df['QB15.6_QB15.7'] = df['QB15.6'] * df['QB15.7']

    cpi_columns = ['CPI 2020', 'CPI 2021', 'CPI 2022', 'CPI 2023']
    if all(col in df.columns for col in cpi_columns):
        df['CPI_avg'] = df[cpi_columns].mean(axis=1)
        for year in ['2020', '2021', '2022', '2023']:
            col_name = f'CPI_{year}_minus_avg'
            df[col_name] = df[f'CPI {year}'] - df['CPI_avg']

        df['CPI_percent_change'] = (df['CPI 2023'] - df['CPI 2020']) / df['CPI 2020']

        df['CPI_category'] = pd.cut(df['CPI_avg'], bins=[0, 50, 70, float('inf')],
                                    labels=['Low', 'Medium', 'High'], right=False)
        category_mapping = {'Low': 0, 'Medium': 1, 'High': 2}
        df['CPI_category_encoded'] = df['CPI_category'].map(category_mapping)

    if 'df_totalscore' in df.columns:
        df['TotalScore_category'] = pd.cut(df['df_totalscore'], bins=[-float('inf'), 5, 10, float('inf')],
                                           labels=['Low', 'Medium', 'High'], right=False)
        score_mapping = {'Low': 0, 'Medium': 1, 'High': 2}
        df['TotalScore_category_encoded'] = df['TotalScore_category'].map(score_mapping)

    return df

def preprocess_eurobarometer(CORRUPTION_DF: pd.DataFrame) -> pd.DataFrame:
    for col in CORRUPTION_DF.columns:
        if col != 'Country':
            CORRUPTION_DF[col] = pd.to_numeric(CORRUPTION_DF[col], errors='coerce')
    
    CORRUPTION_DF = handle_missing_values(CORRUPTION_DF, strategy='mean')
    numerical_columns = ['QB1', 'QB2', 'QB10', 'QB15.1', 'QB15.2', 'QB15.3', 'CPI 2020', 'CPI 2021', 'CPI 2022', 'CPI 2023']
    CORRUPTION_DF = detect_outliers(CORRUPTION_DF, numerical_columns)
    CORRUPTION_DF = add_coordinates(CORRUPTION_DF)
    
    categorical_columns = ['Country']
    CORRUPTION_DF, _ = encode_categorical(CORRUPTION_DF, categorical_columns)
    
    CORRUPTION_DF = create_composite_indices(CORRUPTION_DF)
    CORRUPTION_DF = calculate_additional_features(CORRUPTION_DF)
    
    columns_order = [
            'Country', 'Country_ID', 'QB1', 'QB10', 'QB15.1', 'QB15.10', 'QB15.11', 'QB15.12', 'QB15.13',
            'QB15.2', 'QB15.3', 'QB15.4', 'QB15.5', 'QB15.6', 'QB15.7', 'QB15.8', 'QB15.9',
            'QB4.1', 'QB4.2', 'QB4.3', 'QB4T', 'QB5', 'QB6', 'QB12', 'QB13', 'QB2',
            'QB6_QB13', 'QB15.6_QB15.7', 'QB_avg', 'CPI 2020', 'CPI 2021', 'CPI 2022', 'CPI 2023',
            'CPI_avg', 'CPI_2020_minus_avg', 'CPI_2021_minus_avg', 'CPI_2022_minus_avg',
            'CPI_2023_minus_avg', 'CPI_percent_change', 'CPI_category', 'CPI_category_encoded',
            'TotalScore_category', 'TotalScore_category_encoded', 'df_totalscore', 'df_fulfillment',
            'dj_total score', 'dj_fulfillment', 'tindex_total', 'tindex_fulfillment', 'Latitude', 'Longitude'
    ]
    
    CORRUPTION_DF = CORRUPTION_DF[[col for col in columns_order if col in CORRUPTION_DF.columns]]

    return CORRUPTION_DF

def load_eurobarometer() -> pd.DataFrame:
    CORRUPTION_DATASET_NAME = 'euro-barometer-ebs-a-v1230'
    CORRUPTION_FILE_NAME = 'Corruption_ebs_A_v1230.xlsx'
    CORRUPTION_PATH = f'/kaggle/input/{CORRUPTION_DATASET_NAME}'
    CORRUPTION_FILE_PATH = os.path.join(CORRUPTION_PATH, CORRUPTION_FILE_NAME)
    df = pd.read_excel(os.path.join(CORRUPTION_FILE_PATH), sheet_name=1)
    return preprocess_eurobarometer(df)
    
```

```python
def preprocess_sheet(sheet_name, df):
    """
    Clean the sheets and bring them in an long format
    """
    try:
        # Clean empty rows
        df = df.dropna(how="all").reset_index(drop=True)

        # Setting the header
        columns = ["GEO"] + [f"Year_{i}" for i in range(2013, 2023)] + [f"Unused_{i}" for i in range(2013, 2023)]
        df.columns = columns[:len(df.columns)]

        # Drop irrelevant rows
        df = df.dropna(subset=["GEO"])
        df = df[~df["GEO"].str.contains("TIME|Dataset|Last updated|Unit|Source", na=False)]
        df = df[~df["GEO"].str.match(r"^\d+$", na=False)]

        # Extract every second column
        relevant_columns = ["GEO"] + [col for idx, col in enumerate(df.columns[1:]) if idx % 2 == 0]
        df = df[relevant_columns]

        # Setting the index years
        df.columns = ["GEO"] + [str(year) for year in range(2013, 2013 + len(relevant_columns) - 1)]

        # convert to float
        for col in df.columns[1:]:
            df[col] = pd.to_numeric(df[col], errors="coerce")

        # drop fully empty rows
        df = df.dropna(how="all", subset=df.columns[1:])

        # transform to long fromat
        df = pd.melt(df, id_vars=["GEO"], var_name="Year", value_name="Value")

        # drop empty rows
        df = df.dropna(subset=["Value"])

        return df
    except Exception as e:
        print(f"Error with sheet {sheet_name}: {e}")
        return None


def load_crimes():
    # Inputfile and outputfolder
    input_file = ('/kaggle/input/eurostat-datasets/Recorded-Crimes-until2022.xlsx')
    # Sheets to process
    sheets_to_process = [f"Sheet {i}" for i in range(1, 25, 2)]
    
    # Initializing empty dataframe
    merged_data = pd.DataFrame()
    
    # process every second sheet
    for sheet_name in sheets_to_process:
        try:
            # load sheets
            df = pd.read_excel(input_file, sheet_name=sheet_name, engine="openpyxl", header=None)
    
            # cleant the sheets
            processed_df = preprocess_sheet(sheet_name, df)
    
            if processed_df is not None:
                # include the cleaned sheet to the new outputfile
                merged_data = pd.concat([merged_data, processed_df], ignore_index=True)
        except Exception as e:
            print(f"Error loading Sheet {sheet_name}: {e}")
    
    # test if data is not empty
    if not merged_data.empty:
        # sum up the crimes
        aggregated_data = merged_data.groupby(["GEO", "Year"], as_index=False).sum()
    
        # save the sheet
        output_file = os.path.join("Recorded_Crimes_Total.xlsx")
        aggregated_data.to_excel(output_file, index=False)
    
        # print(f"Cleaned dataset saved to: {output_file}")
    else:
        print("No data found.")
```

```python
input_file = ('/kaggle/input/eurostat-datasets/Educational-Attainment.xlsx')

# relevant sheets
sheets_to_process = [
    'Sheet 19',
    'Sheet 20',
    'Sheet 21',
    'Sheet 22',
    'Sheet 23',
    'Sheet 24'
]

# Mapping
MONTHS_MAPPING = {
    "Jan": "01", "Feb": "02", "Mar": "03", "Apr": "04", "May": "05", "Jun": "06",
    "Jul": "07", "Aug": "08", "Sep": "09", "Oct": "10", "Okt": "10", "Nov": "11", "Dec": "12"
}

# Create a directory for the cleaned sheets
if not os.path.exists(output_folder):
    os.makedirs(output_folder)

def preprocess_sheet(sheet_name, df):
    """
    Preprocess an individual Excel sheet into a clean Pandas DataFrame.
    """
    try:
        # Drop all fully empty rows and columns
        df = df.dropna(how="all").reset_index(drop=True)

        # Find the row containing the header based on the presence of "TIME"
        header_idx = df[df.iloc[:, 0].astype(str).str.contains("TIME", na=False)].index
        if len(header_idx) == 0:
            raise ValueError(f"Header row not found in sheet: {sheet_name}")
        header_idx = header_idx[0]

        # Set the header row and rename columns
        df.columns = df.iloc[header_idx]
        df = df.iloc[header_idx + 1:].reset_index(drop=True)

        # Rename the first column to "Country" for clarity
        df.rename(columns={df.columns[0]: "Country"}, inplace=True)

        # Melt the DataFrame into a long format (Country, Year, Value)
        df = df.melt(id_vars=["Country"], var_name="Year", value_name="Value")

        # Convert Year to numeric where possible
        df["Year"] = pd.to_numeric(df["Year"], errors="coerce")

        # Handle string-based "Value" columns
        def clean_value(value):
            if isinstance(value, str):
                # Replace commas with dots for decimal points
                value = value.replace(",", ".")
                # Split the value if it contains a dot and a potential month
                if "." in value:
                    parts = value.split(".")
                    if len(parts) == 2 and parts[0] in MONTHS_MAPPING:
                        # If the first part matches a month, only keep the numeric part
                        return float(parts[1]) if parts[1].isdigit() else None
                # If it's a pure numeric string, convert to float
                if value.isdigit():
                    return float(value)
            return value

        df["Value"] = df["Value"].apply(clean_value)
        df["Value"] = pd.to_numeric(df["Value"], errors="coerce")

        df = df.dropna(subset=["Country", "Year", "Value"]).reset_index(drop=True)

        # Final cleanup for Country column
        df["Country"] = df["Country"].str.strip()

        # Ensure proper formatting for the "Value" column
        df["Value"] = df["Value"].apply(lambda x: f"{x:.2f}" if isinstance(x, (int, float)) else x)

        # Set a unique index
        df.set_index(["Country", "Year"], inplace=True)

        return df
    except Exception as e:
        print(f"Error processing sheet {sheet_name}: {e}")
        return None

# Process each sheet
for sheet_name in sheets_to_process:
    print(f"Processing sheet: {sheet_name}")
    try:
        # Load the sheet into a Pandas DataFrame
        df = pd.read_excel(input_file, sheet_name=sheet_name, engine="openpyxl")

        # Read the value in cell C9 for naming the output file
        try:
            c9_value = pd.read_excel(input_file, sheet_name=sheet_name, engine="openpyxl", header=None).iloc[8, 2]
            if pd.isna(c9_value):
                raise ValueError("Cell C9 is empty.")
            c9_value = str(c9_value).replace(" ", "_").strip()
        except Exception as e:
            print(f"Error reading cell C9 in sheet {sheet_name}: {e}")
            c9_value = sheet_name  # Fallback to sheet_name if C9 is invalid

        # Preprocess the sheet
        cleaned_df = preprocess_sheet(sheet_name, df)

        # Save the cleaned DataFrame to an Excel file
        if cleaned_df is not None:
            output_path = os.path.join(f"Educ-Attend-{c9_value}_cleaned_.xlsx")

            # Create an Excel workbook
            wb = Workbook()
            ws = wb.active
            ws.title = "Cleaned Data"

            # Format the "Value" column as text
            text_style = NamedStyle(name="text_style")
            text_style.number_format = "@"

            for col in ws.iter_cols(min_col=3, max_col=3):  # Assuming "Value" is the third column
                for cell in col:
                    cell.style = text_style

            # Write data to the Excel sheet
            for row in dataframe_to_rows(cleaned_df.reset_index(), index=False, header=True):
                ws.append(row)

            wb.save(output_path)
            print(f"Saved cleaned sheet: {output_path}")
        else:
            print(f"Skipped sheet due to errors: {sheet_name}")
    except Exception as e:
        print(f"Error reading sheet {sheet_name}: {e}")
```

```python
def basic_prep(df: pd.DataFrame) -> pd.DataFrame:
    # Drop useless columns
    df = df.drop(
        columns=[
            "Region",
            "Country",
            "Rank",
            "Number of sources",
            "Standard error",
            "ISO3",
            "Upper CI",
            "Lower CI",
        ]
    )
    return df


def data_prep(
    df: pd.DataFrame,
    without_year: bool = True,
    row_nan_drop: bool = True,
    europe: bool = True,
) -> pd.DataFrame:
    """
    Prepare the CPI dataset for modeling
    """
    # Consider only WE/EU region
    if europe:
        df = df[df["Region"] == "WE/EU"]
        # Drop columns with more than 100 NaNs
        df = df.drop(columns=df.columns[10:][df.iloc[:, 10:].isna().sum() >= 100])

    if without_year:
        df = df.drop(columns=["Year"])

    if row_nan_drop:
        df = df.dropna()

    return basic_prep(df)


def splitting(df: pd.DataFrame) -> tuple[pd.DataFrame, pd.DataFrame]:
    """
    Split the dataset into features and target
    """
    y = df["CPI score"]
    X = df.drop(columns=["CPI score"])

    return X, y


def train_and_evaluate_model(
    model,
    model_name: str,
    X_train: pd.DataFrame,
    y_train: pd.DataFrame,
    X_test: pd.DataFrame,
    y_test: pd.DataFrame,
):
    """
    Train the model using cross-validation and evaluate on the test set
    """
    # Perform cross-validation
    cv_scores = cross_val_score(
        model, X_train, y_train, cv=5, scoring="neg_mean_squared_error"
    )
    print(f"{model_name} CV   MSE: {-cv_scores.mean()}")

    # Train the model on the full training set
    model.fit(X_train, y_train)

    # Evaluate the model on the test set
    y_pred = model.predict(X_test)
    test_mse = mean_squared_error(y_test, y_pred)
    print(f"{model_name} Test MSE: {test_mse}")
    print()

    return model

def cpi_models(cpi: pd.DataFrame):
    cpi_europe_nonan = data_prep(cpi)
    cpi_europe_nan = data_prep(cpi, row_nan_drop=False)

    cpi_all_nonan = data_prep(cpi, europe=False)
    cpi_all_nan = data_prep(cpi, row_nan_drop=False, europe=False)

    # Compare their shapes and NaN count
    print("Europe nonan shape:", cpi_europe_nonan.shape)
    print("Europe nan shape:", cpi_europe_nan.shape)
    print("All nonan shape:", cpi_all_nonan.shape)
    print("All nan shape:", cpi_all_nan.shape)
    print()
    print("Europe nonan NaNs:", cpi_europe_nonan.isna().sum().sum())
    print("Europe nan NaNs:", cpi_europe_nan.isna().sum().sum())
    print("All nonan NaNs:", cpi_all_nonan.isna().sum().sum())
    print("All nan NaNs:", cpi_all_nan.isna().sum().sum())
    print()

    # Split the datasets
    X_europe_nonan, y_europe_nonan = splitting(cpi_europe_nonan)
    X_europe_nan, y_europe_nan = splitting(cpi_europe_nan)
    X_all_nonan, y_all_nonan = splitting(cpi_all_nonan) # This dataset is empty
    X_all_nan, y_all_nan = splitting(cpi_all_nan)

    # Split the datasets into training and test sets
    X_train, X_test, y_train, y_test = train_test_split(
        X_europe_nonan, y_europe_nonan, test_size=0.2, random_state=SEED
    )

    # Train the linear models
    linear_model = train_and_evaluate_model(
        LinearRegression(), "Linear Regression", X_train, y_train, X_test, y_test
    )
    ridge_model = train_and_evaluate_model(
        Ridge(), "Ridge Regression", X_train, y_train, X_test, y_test
    )
    lasso_model = train_and_evaluate_model(
        Lasso(), "Lasso Regression", X_train, y_train, X_test, y_test
    )

    # Train the gradient boosting model
    gb_model = train_and_evaluate_model(
        HistGradientBoostingRegressor(random_state=SEED, max_iter=300),
        "Gradient Boosting",
        X_train,
        y_train,
        X_test,
        y_test,
    )

    # Train the gradient boosting model with the whole dataset
    X_train_all, X_test_all, y_train_all, y_test_all = train_test_split(
        X_all_nan, y_all_nan, test_size=0.2, random_state=SEED
    )

    gb_model_all = train_and_evaluate_model(
        HistGradientBoostingRegressor(random_state=SEED, max_iter=300),
        "Gradient Boosting with all countries",
        X_train_all,
        y_train_all,
        X_test_all,
        y_test_all,
    )
```

```python
dataset_path = "/kaggle/working/_MergedDatasetQ1andQ4.csv"
dataraw = pd.read_csv(dataset_path)
selected_features = [
    "Unemployment_Rate",
    "Upper_Secondary_Vocational_Education_Level",
    "Upper_Secondary_General_Education_Level",
    "Upper_Secondary_Education_Level",
    "Less_Primary_Education_Level",
    "Tertiary_Education_Level",
    "Crime_Rate",
    "GDP_Per_Capita",
    "mean years of schooling",
    "urban population",
    "population",
    "life expectancy",
]

target_column = "df_fulfillment"
# aggregated features on GEO as more records of features than df_fulfillment
aggregated_features = dataraw.groupby("GEO")[selected_features].mean().reset_index()

# merge with the df_fulfillment
data = pd.merge(aggregated_features, dataraw[["GEO", "df_fulfillment"]].dropna(), on="GEO")
# Define features (X) and target (y)
X = data[selected_features]
y = data[target_column]

# Initialize manual grid search for XGBoost
kf = KFold(n_splits=5, shuffle=True, random_state=42)
best_score = -float("inf")
best_params = None
# Parameter grid for XGBoost
param_grid = {
    "n_estimators": [100, 200, 300],
    "learning_rate": [0.05, 0.1],
    "max_depth": [3, 5],
    "subsample": [0.8, 1.0],
}

# Perform manual cross-validation
for n_estimators in param_grid["n_estimators"]:
    for learning_rate in param_grid["learning_rate"]:
        for max_depth in param_grid["max_depth"]:
            for subsample in param_grid["subsample"]:
                model = XGBRegressor(
                    n_estimators=n_estimators,
                    learning_rate=learning_rate,
                    max_depth=max_depth,
                    subsample=subsample,
                    random_state=42,
                    use_label_encoder=False,
                )
                cv_scores = []
                for train_idx, val_idx in kf.split(X):
                    X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
                    y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]
                    model.fit(X_train, y_train)
                    predictions = model.predict(X_val)
                    cv_scores.append(r2_score(y_val, predictions))
                mean_score = np.mean(cv_scores)
                if mean_score > best_score:
                    best_score = mean_score
                    best_params = {
                        "n_estimators": n_estimators,
                        "learning_rate": learning_rate,
                        "max_depth": max_depth,
                        "subsample": subsample,
                    }
# Train the final model with the best parameters
final_model = XGBRegressor(
    **best_params,
    random_state=42,
)
final_model.fit(X, y)

# Evaluate the final model
predictions = final_model.predict(X)
mse = mean_squared_error(y, predictions)
r2 = r2_score(y, predictions)

# Output results
print("Best Parameters:", best_params)
print("Mean Squared Error:", mse)
print("R² Score:", r2)
```

```python
def eurobarometer_modelling(eurobarometer):    
    # Identify columns related to "perceived corruption" and "actual corruption"
    perceived_corruption_cols = [col for col in eurobarometer.columns if col.startswith('Q') or col == 'cpi']
    actual_corruption_cols = ['df_totalscore', 'df_fulfillment', 'dj_total score', 'dj_fulfillment', 'tindex_total', 'tindex_fulfillment']
    
    # Separate data for perceived and actual corruption
    perceived_corruption_data = eurobarometer[perceived_corruption_cols]
    actual_corruption_data = eurobarometer[actual_corruption_cols]
    
    # Align years for CPI and corruption data (if a "year" column exists)
    if 'year' in eurobarometer.columns:
        perceived_corruption_data = perceived_corruption_data[eurobarometer['year'] == 2023]
        actual_corruption_data = actual_corruption_data[eurobarometer['year'] == 2023]
    
    # Handle missing values
    perceived_corruption_data = perceived_corruption_data.fillna(0)
    actual_corruption_data = actual_corruption_data.fillna(0)
    
    # Scale the data
    scaler = StandardScaler()
    perceived_scaled = scaler.fit_transform(perceived_corruption_data)
    actual_scaled = scaler.fit_transform(actual_corruption_data)
    
    # Clustering to find patterns in perceived and actual corruption
    kmeans_perceived = KMeans(n_clusters=3, random_state=42)
    kmeans_actual = KMeans(n_clusters=3, random_state=42)
    
    # Fit clustering directly on scaled data without PCA
    perceived_clusters = kmeans_perceived.fit_predict(perceived_scaled)
    actual_clusters = kmeans_actual.fit_predict(actual_scaled)
    
    perceived_labels = {0: "Low Perceived Corruption", 1: "Moderate Perceived Corruption", 2: "High Perceived Corruption"}
    actual_labels = {0: "Low Actual Corruption", 1: "Moderate Actual Corruption", 2: "High Actual Corruption"}
    
    # Visualize patterns directly using original scaled data
    plt.figure(figsize=(12, 6))
    
    # Visualization for perceived corruption clusters
    plt.subplot(1, 2, 1)
    scatter1 = plt.scatter(perceived_scaled[:, 0], perceived_scaled[:, 1], c=perceived_clusters, cmap='viridis')
    plt.title('Perceived Corruption Clusters')
    plt.xlabel('Feature 1 (Scaled)')
    plt.ylabel('Feature 2 (Scaled)')
    plt.legend(handles=scatter1.legend_elements()[0], labels=list(perceived_labels.values()), title="Clusters")
    
    # Visualization for actual corruption clusters
    plt.subplot(1, 2, 2)
    scatter2 = plt.scatter(actual_scaled[:, 0], actual_scaled[:, 1], c=actual_clusters, cmap='viridis')
    plt.title('Actual Corruption Clusters')
    plt.xlabel('Feature 1 (Scaled)')
    plt.ylabel('Feature 2 (Scaled)')
    plt.legend(handles=scatter2.legend_elements()[0], labels=list(actual_labels.values()), title="Clusters")
    
    plt.tight_layout()
    plt.show()
    
    # Cross-analysis between perceived and actual corruption
    correlation_matrix = pd.DataFrame(
        np.corrcoef(perceived_scaled.T, actual_scaled.T)[:len(perceived_corruption_cols), len(perceived_corruption_cols):],
        index=perceived_corruption_cols,
        columns=actual_corruption_cols
    )
    
    print("Correlation Matrix between Perceived and Actual Corruption:")
    print(correlation_matrix)
    # QUESTION 2 : Differences between perceived and actual corruption
    # Add cluster differences and correlation analysis
    cluster_diff = pd.DataFrame({
        'Perceived Cluster': perceived_clusters,
        'Actual Cluster': actual_clusters
    })
    print("Cluster Differences between Perceived and Actual Corruption:")
    print(cluster_diff.value_counts())
    
    # Create a heatmap for better visualization of cluster combinations
    cluster_diff_table = cluster_diff.value_counts().reset_index()
    cluster_diff_table.columns = ['Perceived Cluster', 'Actual Cluster', 'Count']
    
    # Pivot the table for heatmap visualization
    heatmap_data = cluster_diff_table.pivot(index="Perceived Cluster", columns="Actual Cluster", values="Count")
    
    # Plot the heatmap
    plt.figure(figsize=(8, 6))
    sns.heatmap(heatmap_data, annot=True, cmap="coolwarm", fmt='.2f', cbar_kws={'label': 'Count'})
    plt.title("Cluster Differences Between Perceived and Actual Corruption")
    plt.xlabel("Actual Cluster")
    plt.ylabel("Perceived Cluster")
    plt.show()
    
    # Create target variable: Actual corruption level (binary classification for simplicity)
    eurobarometer['Target'] = (eurobarometer['df_totalscore'] > eurobarometer['df_totalscore'].median()).astype(int)  # High corruption = 1, Low corruption = 0
    
    # Features: Include all enriched features, including CPI differences, interactions, and encodings
    features = [
        'QB1', 'QB10', 'QB15.1', 'QB15.10', 'QB15.11', 'QB15.12', 'QB15.13',
        'QB15.2', 'QB15.3', 'QB15.4', 'QB15.5', 'QB15.6', 'QB15.7', 'QB15.8', 'QB15.9',
        'QB4.1', 'QB4.2', 'QB4.3', 'QB4T', 'QB5', 'QB6', 'QB12', 'QB13', 'QB2',
        'QB6_QB13', 'QB15.6_QB15.7', 'QB_avg',
        'CPI_2020_minus_avg', 'CPI_2021_minus_avg', 'CPI_2022_minus_avg', 'CPI_2023_minus_avg',
        'CPI_percent_change', 'CPI_category_encoded', 'TotalScore_category_encoded',
        
    ]
    
    X = eurobarometer[features]
    y = eurobarometer['Target']
    
    # Handle missing values
    X = X.fillna(0)
    
    # Split data into training and testing sets
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
    
    # Step 1: Hyperparameter Tuning for Random Forest
    param_grid = {
        'n_estimators': [50, 100, 200],
        'max_depth': [None, 10, 20, 30],
        'min_samples_split': [2, 5, 10],
        'min_samples_leaf': [1, 2, 4]
    }
    grid_search = GridSearchCV(RandomForestClassifier(random_state=42), param_grid, cv=5, scoring='accuracy', n_jobs=-1)
    grid_search.fit(X_train, y_train)
    
    # Best Random Forest model
    best_rf_model = grid_search.best_estimator_
    
    # Evaluate the tuned model
    best_predictions = best_rf_model.predict(X_test)
    best_conf_matrix = confusion_matrix(y_test, best_predictions)
    best_class_report = classification_report(y_test, best_predictions)
    
    # Step 2: Feature Importance Visualization
    feature_importances = pd.DataFrame({
        'Feature': X_train.columns,
        'Importance': best_rf_model.feature_importances_
    }).sort_values(by='Importance', ascending=False)
    
    plt.figure(figsize=(10, 6))
    sns.barplot(x=feature_importances['Importance'], y=feature_importances['Feature'])
    plt.title("Feature Importances")
    plt.xlabel("Importance")
    plt.ylabel("Feature")
    plt.tight_layout()
    plt.show()
    
    # Step 3: Confusion Matrix Visualization
    plt.figure(figsize=(6, 5))
    sns.heatmap(best_conf_matrix, annot=True, fmt="d", cmap="Blues", xticklabels=["Low Corruption", "High Corruption"], yticklabels=["Low Corruption", "High Corruption"])
    plt.title("Confusion Matrix")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.tight_layout()
    plt.show()
    
    # Step 4: Comparing Perceived vs Actual Corruption
    # Use most important feature for visualization
    most_important_feature = feature_importances.iloc[0]['Feature']
    plt.figure(figsize=(8, 6))
    sns.scatterplot(x=eurobarometer[most_important_feature], y=eurobarometer['df_totalscore'], hue=y, palette='coolwarm')
    plt.title(f"Perceived vs Actual Corruption (Feature: {most_important_feature})")
    plt.xlabel(most_important_feature)
    plt.ylabel("Actual Corruption Score")
    plt.legend(title="Actual Corruption Level", loc="best")
    plt.tight_layout()
    plt.show()
    
    # Display results
    print("Best Random Forest Model Classification Report:")
    print(best_class_report)
    
    # --- QUESTION 3: Different forms of corruption across countries in Europe
    
    # Add perceived and actual clusters to the dataset
    eurobarometer['Perceived Cluster'] = perceived_clusters
    eurobarometer['Actual Cluster'] = actual_clusters
    
    # Count the frequency of each cluster for each country
    perceived_cluster_counts = eurobarometer.groupby('Country')['Perceived Cluster'].value_counts().unstack(fill_value=0)
    actual_cluster_counts = eurobarometer.groupby('Country')['Actual Cluster'].value_counts().unstack(fill_value=0)
    
    # Perceived corruption clusters table
    print("Perceived Corruption Cluster Counts by Country (Table):")
    perceived_cluster_table = perceived_cluster_counts.reset_index()
    display(perceived_cluster_table)
    
    # Actual corruption clusters table
    print("\nActual Corruption Cluster Counts by Country (Table):")
    actual_cluster_table = actual_cluster_counts.reset_index()
    display(actual_cluster_table)
    
    # Calculate discrepancies between perceived and actual corruption clusters for each country
    eurobarometer['Cluster Discrepancy'] = eurobarometer['Perceived Cluster'] - eurobarometer['Actual Cluster']
    
    # Group by country to calculate the mean discrepancy and summarize the results
    discrepancy_table = eurobarometer.groupby('Country')['Cluster Discrepancy'].mean().reset_index()
    discrepancy_table.columns = ['Country', 'Mean Cluster Discrepancy']
    
    # Display the discrepancy table
    print("Discrepancy Table: Mean Differences Between Perceived and Actual Corruption Clusters by Country")
    display(discrepancy_table)
    
    # Identify variables contributing to discrepancies
    # Calculate the correlation between features and the discrepancy column
    correlation_with_discrepancy = eurobarometer[perceived_corruption_cols + ['Cluster Discrepancy']].corr()['Cluster Discrepancy'].drop('Cluster Discrepancy').sort_values(ascending=False)
    
    # Display the top variables contributing to discrepancies
    print("\nTop Variables Contributing to Cluster Discrepancies:")
    correlation_table = correlation_with_discrepancy.reset_index()
    correlation_table.columns = ['Feature', 'Correlation with Discrepancy']
    display(correlation_table)
    
    # -- Conclusions
    
    # Target variable: Corruption level
    eurobarometer['Target'] = (eurobarometer['df_totalscore'] > eurobarometer['df_totalscore'].median()).astype(int)
    
    perceived_corruption_cols = [col for col in eurobarometer.columns if col.startswith('QB')]
    cpi_columns = ['CPI 2020', 'CPI 2021', 'CPI 2022', 'CPI 2023']
    
    X = eurobarometer[perceived_corruption_cols + cpi_columns].fillna(0)
    
    y = eurobarometer['Target']
    
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
    
    rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
    rf_model.fit(X_train, y_train)
    
    predictions = rf_model.predict(X_test)
    
    conf_matrix = confusion_matrix(y_test, predictions)
    class_report = classification_report(y_test, predictions)
    
    feature_importances = pd.DataFrame({
        'Feature': X.columns,
        'Importance': rf_model.feature_importances_
    }).sort_values(by='Importance', ascending=False)
    
    print("Confusion Matrix:")
    print(conf_matrix)
    print("\nClassification Report:")
    print(class_report)
    print("\nFeature Importances:")
    print(feature_importances)
    
    plt.figure(figsize=(10, 6))
    sns.barplot(x=feature_importances['Importance'], y=feature_importances['Feature'])
    plt.title("Feature Importances (Perceived Metrics and Historical CPI)")
    plt.xlabel("Importance")
    plt.ylabel("Feature")
    plt.tight_layout()
    plt.show()
    
    plt.figure(figsize=(6, 5))
    sns.heatmap(conf_matrix, annot=True, fmt="d", cmap="Blues", xticklabels=["Low Corruption", "High Corruption"], yticklabels=["Low Corruption", "High Corruption"])
    plt.title("Confusion Matrix")
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.tight_layout()
    plt.show()

```
