```python
def preprocess_sheet(sheet_name, df):
    """
    Preprocesses a single sheet by cleaning, reshaping, and standardizing data.
    """
    print(f"Processing sheet: {sheet_name}")
    try:
        # Drop rows with all NaN values
        df.dropna(how='all', inplace=True)

        # Ensure all column headers are strings
        df.columns = df.columns.astype(str)

        # Identify the header row containing "TIME" and reset column names
        header_idx = df.index[df.iloc[:, 0].astype(str).str.contains("TIME", na=False)].tolist()
        if header_idx:
            header_idx = header_idx[0]
            df.columns = df.iloc[header_idx].astype(str)
            df = df[header_idx + 1:].reset_index(drop=True)

        # Remove unnecessary columns (columns named "Unnamed")
        df = df.loc[:, ~df.columns.str.contains('^Unnamed', na=False)]

        # Standardize column names
        df.columns = [str(col).replace(' ', '_').strip() for col in df.columns]

        # Dynamically identify the column containing country names
        country_col = next((col for col in df.columns if 'GEO' in col.lower() or 'country' in col.lower()), None)
        if not country_col:
            raise ValueError(f"Country column not found in sheet {sheet_name}.")

        # Replace specific country names for consistency
        df[country_col] = df[country_col].astype(str).replace({
            'European Union - 27 countries (from 2020)': 'EuropeanUnion',
            'Euro area – 20 countries (from 2023)': 'EuroArea'
        })

        # Reshape the dataframe into a long format
        id_vars = [country_col]
        value_vars = [col for col in df.columns if col not in id_vars]
        df_long = pd.melt(df, id_vars=id_vars, value_vars=value_vars, 
                          var_name='Year', value_name='Value')

        # Rename columns for consistency
        df_long.rename(columns={country_col: 'Country'}, inplace=True)

        # Convert 'Year' and 'Value' columns to numeric types
        df_long['Year'] = pd.to_numeric(df_long['Year'], errors='coerce')
        df_long['Value'] = pd.to_numeric(df_long['Value'], errors='coerce')

        # Drop rows with missing values in 'Year' or 'Value'
        df_long.dropna(subset=['Year', 'Value'], inplace=True)

        # Return the cleaned dataframe
        return df_long
    except Exception as e:
        print(f"Error during preprocessing for sheet {sheet_name}: {e}")
        return None
```

```python
def process_and_save(input_excel, sheet_names, output_dir):
    """
    Process all given sheets from the input Excel file and save cleaned outputs to the specified directory.
    """
    for sheet_name in sheet_names:
        try:
            # Read the sheet from the Excel file
            df = pd.read_excel(input_excel, sheet_name=sheet_name)

            # Preprocess the sheet
            cleaned_df = preprocess_sheet(sheet_name, df)

            if cleaned_df is not None:
                # Save the cleaned dataframe as a CSV file with ; delimiter
                output_path = f"{output_dir}/{sheet_name}_formatted.csv"
                cleaned_df.to_csv(output_path, index=False, sep=';')
                print(f"Saved cleaned sheet to: {output_path}")
        except Exception as e:
            print(f"Error processing sheet {sheet_name}: {e}")
```

```python
# Process and save the sheets
process_and_save(input_excel, sheet_names, output_dir)
```
